```python
FRIENDLY_BOTS = {
    # Search engines
    "googlebot": [
        "googlebot",
    ],
    "bingbot": [
        "bingbot",
    ],

    # Google Ads / AdSense
    "mediapartners_google": [
        "mediapartners-google",
    ],

    # OpenAI / AI search you want to keep
    "gptbot": [
        "gptbot",
    ],
    "oai_searchbot": [
        "oai-searchbot",
        "openai.com/searchbot",
    ],

    # Ad partners / verification useful for ads ecosystem
    "criteobot": [
        "criteobot",
        "criteo-crawler",
    ],
    "ias": [
        "ias-va",
        "ias-or",
        "ias crawler",
        "ias_crawler",
        "integralads",
        "admantx",
    ],
}

agents = view_drains['proxy.userAgent'].unique()
```

```python
view_stats = (
    view_drains
    .groupby(['proxy.userAgent'])
    .agg(
        conteo_routes_visitadas = ('routes_visited', 'sum')
    )
)

view_stats = view_stats.sort_values('conteo_routes_visitadas', ascending = False)
```

```python
view_stats
```

```python


def detect_friendly_bot(user_agent, bot_patterns):

    ua = str(user_agent).lower()

    for bot_name, patterns in bot_patterns.items():

        for pattern in patterns:

            if pattern.lower() in ua:

                return bot_name

    return None

view_stats_labeled = view_stats.copy()

view_stats_labeled["friendly_bot_type"] = view_stats_labeled.index.map(

    lambda ua: detect_friendly_bot(ua, FRIENDLY_BOTS)

)

view_stats_labeled["is_friendly"] = view_stats_labeled["friendly_bot_type"].notna()

view_stats_labeled = view_stats_labeled.sort_values(

    "conteo_routes_visitadas",

    ascending=False

)

view_stats_labeled
```
